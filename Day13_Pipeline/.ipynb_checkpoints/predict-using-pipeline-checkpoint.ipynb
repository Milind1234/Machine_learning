# Machine Learning Pipelines in Sklearn

# What is an ML Pipeline?
# ------------------------
# A pipeline in sklearn is a way to chain multiple preprocessing and modeling steps into a single object.
# It automates and ensures that the output from one step (e.g., imputation) is the input for the next
# (e.g., encoding, then modeling) — keeping everything organized, less error-prone, and ready for deployment.

from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_selection import SelectKBest, f_classif

# Typical Workflow:
# 1. Raw data (may include missing values, categorical variables, irrelevant columns)
# 2. Preprocessing steps: missing value imputation, encoding, scaling, feature selection, etc.
# 3. Model fitting (training)
# 4. Evaluation, export, and deployment
# Pipelines let us "bundle" these steps.

# Example: Titanic dataset (common for introductory ML tasks)

# Drop irrelevant columns
drop_cols = ['PassengerId', 'Name', 'Ticket', 'Cabin']
# train_test_split and other imports omitted for brevity

# Step 1: Preprocessing transformers
# Imputation for numerical columns
num_imputer = SimpleImputer(strategy='mean')
# Imputation for categorical columns
cat_imputer = SimpleImputer(strategy='most_frequent')
# OneHotEncoder for string categorical columns
ohe = OneHotEncoder(handle_unknown='ignore', sparse=False)

# Step 2: Bundle transformers with ColumnTransformer
# Example column indices/names (update as per your dataframe)
num_cols = [0, 5]             # e.g., Age, Fare
cat_cols = [1, 6]             # e.g., Sex, Embarked

preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_imputer, num_cols),
        ('cat', cat_imputer, cat_cols)
    ],
    remainder='passthrough'
)

# Step 3: Final pipeline construction (with modeling)
pipeline = Pipeline(steps=[
    ('preprocessing', preprocessor),
    ('encoding', OneHotEncoder(handle_unknown='ignore', sparse=False)),
    # Optionally, add feature selection:
    # ('feature_selection', SelectKBest(score_func=f_classif, k=5)),
    ('model', DecisionTreeClassifier())
])

# Benefits of using pipelines:
# - Reproducible and consistent preprocessing for training and testing
# - Clean and modular code
# - Easy deployment (delay preprocessing logic — now model expects raw input)
# - Facilitates hyperparameter tuning and cross-validation on the *entire* workflow

# - Pipeline: More explicit; you must name each step.
# - make_pipeline: Step names autogenerated; best for quick prototyping.

# - Use Pipeline when step-level referencing or CV tuning is needed.
