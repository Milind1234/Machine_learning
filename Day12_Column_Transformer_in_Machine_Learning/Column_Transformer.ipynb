# Column Transformer in Machine Learning - Sklearn

# Problem:
# Different columns in your dataset require different preprocessing steps.
# Example: Age (numeric, impute missing), City & Gender (nominal categorical, one-hot encode),
#          Review (ordinal categorical, ordinal encode)
# Usually, applying transformations individually can be messy and error-prone.

# Solution:
# ColumnTransformer class in sklearn allows you to apply specific transformations to specific columns,
# and then combines the results into a single output.

# Sample Dataset:
# Columns: age (numeric), city (nominal), gender (nominal), review (ordinal)
# Each column requires a different transformation.

# Step 1: Import Necessary Classes
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.impute import SimpleImputer

# Step 2: Define the transformers for each type of column

# Example: Suppose you have these columns
# Numerical columns (age, fever): Need missing value imputation
# Ordinal categorical (review): Requires OrdinalEncoder
# Nominal categorical (city, gender): Requires OneHotEncoder

# For missing value imputation (numerical)
num_transformer = SimpleImputer(strategy='mean')

# For ordinal categorical encoding (reviews: ['bad', 'average', 'good', 'excellent'])
ord_transformer = OrdinalEncoder(categories=[['bad', 'average', 'good', 'excellent']])

# For nominal categorical encoding (city, gender)
nom_transformer = OneHotEncoder(drop='first', sparse=False)

# Step 3: Use ColumnTransformer to tie everything together
preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_transformer, ['age', 'fever']),
        ('ord', ord_transformer, ['review']),
        ('nom', nom_transformer, ['city', 'gender'])
    ],
    remainder='drop'  # Drop any columns not specified above
)

# Step 4: Fit and transform your data
# X_train is your input DataFrame
# X_preprocessed = preprocessor.fit_transform(X_train)

# Advantages:
# - Apply different preprocessing to different columns in one pipeline
# - Easily scalable to dozens of columns and transformations
# - Keeps code organized, less error-prone

# Additional Notes:
# - remainder='passthrough' can be used to keep untouched columns.
# - Each transformer tuple: (<name>, <transformer_obj>, <list_of_columns>)
# - ColumnTransformer helps when building robust ML pipelines.

# Example usage:
# from sklearn.pipeline import Pipeline
# full_pipeline = Pipeline(steps=[
#     ('preprocessor', preprocessor),
#     ('classifier', SomeClassifier())

# 1. Without Column Transformer

```python
#adding simple imputer to fever col
si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[['fever']])

#also the test data
X_test_fever = si.fit_transform(X_test[['fever']])
X_train_fever.shape
```

```python
#OrdinalEncoding -> cough
oe= OrdinalEncoder(categories = [['Mild','Strong']])
X_train_cough = oe.fit_transform(X_train[['cough']])

#also the test data
X_test_cough = oe.fit_transform(X_test[['cough']])

X_train_cough.shape
```

```python
#onehotencoding -> gender , city
ohe = OneHotEncoder(drop = 'first',sparse_output = False,dtype=np.int32)
X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])

#also the test data
X_test_gender_city = ohe.fit_transform(X_test[['gender','city']])

X_train_gender_city
```

```python
# extracting age by droping other cols
X_train_age = X_train.drop(columns = ['gender','fever','cough','city']).values

#also for test data
X_test_age = X_test.drop(columns = ['gender','fever','cough','city']).values

X_train_age.shape
```

```python
X_train_transformed = np.concatenate((X_train_age,X_train_fever,X_train_gender_city,X_train_cough),axis =1)
X_test_transformed = np.concatenate((X_test_age,X_test_fever,X_test_gender_city,X_test_cough),axis =1)
X_train_transformed
```

# 2. Using ColumnTransformer

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers = [
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories =[['Mild','Strong']]),['cough']),
    ('tnf3',OneHotEncoder(sparse_output = False,drop='first'),['gender','city'])
],remainder = 'passthrough')
```

```python
transformer.fit_transform(X_train)
```

```python
transformer.transform(X_test)
```
